```python
from collections import Counter
grades = [83, 95, 91, 87, 70, 0, 85, 82, 100, 67, 73, 77, 0]

histogram = Counter(min(grade // 10*10, 90) for grade in grades)

plt.bar([x + 5 for x in histogram.keys()],
        histogram.values(),
        10,
        edgecolor=(0,0,0))
plt.axis([-5, 105, 0, 5])

plt.xticks([10 * i for i in range(11)])
plt.xlabel("Decile")
plt.ylabel("# of Students")
plt.title("Distribution of Exam 1 Grades")
plt.show()
```

```python
variance = [1, 2, 4, 8, 16, 32, 64, 128, 256]
bias_squared = [256, 128, 64, 32, 16, 8, 4, 2, 1]
total_error = [x + y for x,y in zip(variance, bias_squared)]
xs = [i for i, _ in enumerate(variance)]

plt.plot(xs, variance, 'g-', label='variance')
plt.plot(xs, bias_squared, 'r-.', label='bias^2')
plt.plot(xs, total_error, 'b:', label='total error')

plt.legend(loc=9)
plt.xlabel("model complexity")
#plt.xticks([])
plt.title("The Bias-Variance Tradeoff")
plt.show()
```

```python
friends = [70, 65, 72, 63, 71, 64, 60, 64, 67]
minutes = [175, 170, 205, 120, 220, 130, 105, 145, 190]
labels = ['a','b','c','d','e','f','g','h','i']

plt.scatter(friends, minutes)

for label, friend_count, minute_count in zip(labels, friends, minutes):
  plt.annotate(label,
               xy =(friend_count, minute_count),
               xytext=(5,-5),
               textcoords='offset points')

plt.title("Daily Minutes vs. Number of Friends")
plt.xlabel("# of friends")
plt.ylabel("daily minutes spent on the site")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'scores': [56, 67, 45, 89, 90, 72, 78, 88, 94, 60, 73, 81, 85, 68, 77]
})

data['scores'].hist(bins=8, edgecolor='black')

plt.title("Histogram of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
import seaborn as sns

sns.histplot(data['scores'], bins=8, kde=True)  # kernel density estimate
plt.title("Histogram of Exam Scores with KDE Curve")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 200)
plt.figure()

plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')

plt.title('Sine & Cosine')
plt.xlabel('x') 
plt.ylabel('y')

plt.legend()
plt.show()
```

```python
#Task: Given categories ['A','B','C','D'] and counts [12,7,19,5], make a bar chart with title and axis labels.
```

```python
categories = ['A', 'B', 'C', 'D']
counts = [12, 7, 19, 5]

plt.figure(figsize=(8, 6))
plt.bar(categories, counts, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])

plt.title('Category Counts', fontsize=14, fontweight='bold')
plt.xlabel('Categories', fontsize=12)
plt.ylabel('Counts', fontsize=12)

for i, count in enumerate(counts):
    plt.text(i, count + 0.3, str(count), ha='center', va='bottom')

plt.ylim(0, max(counts) + 2)

plt.tight_layout()
plt.show()
```

```python
#Task: Generate 1000 samples from Normal(0,1). Plot a histogram with 30 bins and clear axis labels.
```

```python
data = np.random.normal(loc=0, scale=1, size=1000)
plt.figure()
plt.hist(data, bins=30, edgecolor='black')
plt.title('Normal(0,1) Histogram')
plt.xlabel('Value'); plt.ylabel('Frequency')
plt.show()
```

```python
#Task: Create x from 0 to 10(80 points) and  y = 0.8x + noise. Plot a scatter; add title and labels
```

```python
x = np.linspace(0, 10, 80)
y = 0.8*x + np.random.normal(0, 1, size=x.size)
plt.figure()
plt.scatter(x, y)
plt.title('Scatter with Linear Trend')
plt.xlabel('x'); plt.ylabel('y')
plt.show()
```

```python
#Task: Draw a pie chart for labels = ['North','South','East','West'] and shares = [40,25,20,15] with percentages.
```

```python
labels = ['North','South','East','West']
shares = [40, 25, 20, 15]

plt.figure()
plt.pie(shares, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title('Regional Share')
plt.axis('equal')
plt.show()
```
